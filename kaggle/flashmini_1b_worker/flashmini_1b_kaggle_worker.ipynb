```python
import os, runpy, sys
from pathlib import Path
os.environ.setdefault('FLASHMINI_REMOTE_CHECKPOINT_DATASET', "masterjaso/flashmini-1b-checkpoints")
bundle_roots = [Path.cwd(), Path('/kaggle/working')] + list(Path('/kaggle/input').glob('*'))
bundle = next((root for root in bundle_roots if (root / 'flashmini_1b_kaggle_worker.py').is_file()), None)
if bundle is None:
    raise RuntimeError('FlashMini worker bundle is not mounted under /kaggle/input or /kaggle/working')
sys.path.insert(0, str(bundle / 'src'))
sys.argv = ['flashmini_1b_kaggle_worker.py', '--run-dir', '/kaggle/working/flashmini_run', '--freeze-manifest', str(bundle / 'freeze_manifest.json'), '--smoke']
runpy.run_path(str(bundle / 'flashmini_1b_kaggle_worker.py'), run_name='__main__')
```
